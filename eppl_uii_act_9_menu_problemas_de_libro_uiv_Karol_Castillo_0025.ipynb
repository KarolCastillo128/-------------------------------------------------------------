```python
def vermenu():
    print("\n--- MENU ---")
    print("1. Becas a estudiantes (3.7)")
    print("2. Bono de trabajador (3.8)")
    print("3. Consultas medicas (3.12)")
    print("4. Paqueteria (3.14)")
    print("5. Precio de articulo (3.16)")
    print("6. Salir")
    opcion = int(input("Ingrese la opcion: "))
    return opcion

def verificarSigno(valor):
    if valor >= 0:
        print("El resultado es POSITIVO")
    else:
        print("El resultado es NEGATIVO")

# 3.7
def Becas():
    edad = int(input("Ingrese edad: "))
    prom = float(input("Ingrese promedio: "))

    if edad > 18:
        if prom >= 9:
            beca = 2000
        elif prom >= 7.5:
            beca = 1000
        elif prom >= 6:
            beca = 500
        else:
            beca = 0
            print("Carta de invitacion")
    else:
        if prom >= 9:
            beca = 3000
        elif prom >= 8:
            beca = 2000
        elif prom >= 6:
            beca = 100
        else:
            beca = 0
            print("Carta de invitacion")

    print("Beca:", beca)
    verificarSigno(beca)

# 3.8
def BonoTrabajador():
    ant = int(input("Ingrese antiguedad: "))
    sueldo = float(input("Ingrese sueldo: "))

    if ant > 2 and ant < 5:
        bono1 = sueldo * 0.20
    elif ant >= 5:
        bono1 = sueldo * 0.30
    else:
        bono1 = 0

    if sueldo < 1000:
        bono2 = sueldo * 0.25
    elif sueldo <= 3500:
        bono2 = sueldo * 0.15
    else:
        bono2 = sueldo * 0.10

    if bono1 > bono2:
        bono = bono1
    else:
        bono = bono2

    print("Bono:", bono)
    verificarSigno(bono)

# 3.12
def ConsultasMedicas():
    nc = int(input("Ingrese numero de consultas: "))

    if nc <= 3:
        total = nc * 200
    elif nc <= 5:
        total = (nc - 3) * 150 + 600
    elif nc <= 8:
        total = (nc - 5) * 100 + 900
    else:
        total = (nc - 8) * 50 + 1200

    print("Total a pagar:", total)
    verificarSigno(total)

# 3.14
def Paqueteria():
    zona = int(input("Ingrese zona (1-5): "))
    peso = float(input("Ingrese peso en gramos: "))

    if peso > 5000:
        print("No se puede enviar")
        total = 0
    else:
        if zona == 1:
            total = peso * 11
        elif zona == 2:
            total = peso * 10
        elif zona == 3:
            total = peso * 12
        elif zona == 4:
            total = peso * 24
        elif zona == 5:
            total = peso * 27
        else:
            total = 0
            print("Zona invalida")

        print("Costo:", total)

    verificarSigno(total)

# 3.16
def PrecioArticulo():
    clave = int(input("Ingrese clave del articulo: "))
    mp = float(input("Ingrese materia prima: "))

    if clave == 3 or clave == 4:
        mo = mp * 0.75
    elif clave == 1 or clave == 5:
        mo = mp * 0.80
    else:
        mo = mp * 0.85

    if clave == 2 or clave == 5:
        gf = mp * 0.30
    elif clave == 3 or clave == 6:
        gf = mp * 0.35
    else:
        gf = mp * 0.28

    cp = mp + mo + gf
    pv = cp + (cp * 0.45)

    print("Costo de produccion:", cp)
    print("Precio de venta:", pv)
    verificarSigno(pv)

# MENU PRINCIPAL
while True:
    opcion = vermenu()

    if opcion == 1:
        Becas()
    elif opcion == 2:
        BonoTrabajador()
    elif opcion == 3:
        ConsultasMedicas()
    elif opcion == 4:
        Paqueteria()
    elif opcion == 5:
        PrecioArticulo()
    elif opcion == 6:
        print("Gracias por usar el programa")
        print("programa realizado por Karol Castillo = 9")
        break
    else:
        print("Opcion invalida")
```
