#### problema 7+3= 10
Conversión de Divisas (Monedas) Simula un tipo de cambio (ej. basado en dólar)
1. USD a MXN (Peso Mexicano)
2. USD a EUR (Euro)
3. USD a COP (Peso Colombiano)
4. Salir.

```python
# problema No 7+3 = 10 (Conversión de Divisas)
# menu de opciones (1 USD a MXN, 2 USD a EUR, 3 USD a COP, 4 salir)
# usando if, while true, break y entrada de datos
# llamando a funciones usd_mxn(), usd_eur(), usd_cop(), menu (), captura datos()

def menu():
    print("menu de opciones")
    print("1 USD a MXN (Peso Mexicano)")
    print("2 USD a EUR (Euro)")
    print("3 USD a COP (Peso Colombiano)")
    print("4 salir")
    opcion = int(input("Ingrese una opcion: "))
    return opcion

def captura_datos():
    num1 = float(input("Ingrese la cantidad en USD: "))
    num2 = float(input("Ingrese el tipo de cambio: "))
    return num1, num2

def usd_mxn(cantidad, tipo_cambio):
    if tipo_cambio == 0:
        print("Error: El tipo de cambio no puede ser cero.")
    else:
        resultado = cantidad * tipo_cambio
        print("El valor en MXN es:", resultado)

def usd_eur(cantidad, tipo_cambio):
    if tipo_cambio == 0:
        print("Error: El tipo de cambio no puede ser cero.")
    else:
        resultado = cantidad * tipo_cambio
        print("El valor en EUR es:", resultado)

def usd_cop(cantidad, tipo_cambio):
    if tipo_cambio == 0:
        print("Error: El tipo de cambio no puede ser cero.")
    else:
        resultado = cantidad * tipo_cambio
        print("El valor en COP es:", resultado)

while True:
    opcion = menu() # llama a funcion menu
    if opcion == 1:
        num1, num2 = captura_datos()
        usd_mxn(num1, num2) # llama a funcion usd_mxn
    elif opcion == 2:
        num1, num2 = captura_datos()
        usd_eur(num1, num2) # llama a funcion usd_eur
    elif opcion == 3:
        num1, num2 = captura_datos()
        usd_cop(num1, num2) # llama funcion usd_cop
    elif opcion == 4:
        break
    else:
        print("Opcion invalida")

print("Fin del programa")
print("Programa realizado por Emily Barraza NL = 7")
```
